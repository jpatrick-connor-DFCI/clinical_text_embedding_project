# Render manuscript figures (R tier)

Runs the five `R/plot_figure_*.R` scripts in this notebook's **R kernel** (`IRkernel`). Each
script reads the CSVs written by the Python prep tier and composes its figure in-memory with
`ggplot2 + patchwork`, saving `figureN_*.{png,pdf}` to `$CLINICAL_FIGURES_OUT/target_figures/`
plus individual panel PNGs alongside.

- **Prep tier**: [generate_figure_data.ipynb](generate_figure_data.ipynb) (Python kernel) — run first.
- **One-time R package bootstrap** (any host with R):

  ```
  Rscript jupyter_notebooks/manuscript_figures/R/install_packages.R
  ```

Each plot script is `sys.source()`-ed in its own environment so per-script ggplot objects
(`p2a`, `p3b`, …) don't leak between cells.

````r
find_repo_root <- function() {
  d <- normalizePath(getwd(), mustWork = TRUE)
  while (nchar(d) > 1) {
    if (dir.exists(file.path(d, "python_scripts")) &&
        dir.exists(file.path(d, "jupyter_notebooks"))) return(d)
    parent <- dirname(d)
    if (parent == d) break
    d <- parent
  }
  stop("Could not find repo root from ", getwd())
}

REPO_ROOT <- find_repo_root()
R_DIR     <- file.path(REPO_ROOT, "jupyter_notebooks", "manuscript_figures", "R")

cat("Repo root: ", REPO_ROOT, "\n", sep = "")
cat("R scripts: ", R_DIR, "\n", sep = "")
cat("R version: ", R.version.string, "\n", sep = "")
````

````r
PLOT_SCRIPTS <- c(
  "plot_figure_1.R",
  "plot_figure_2.R",
  "plot_figure_2_supp.R",
  "plot_figure_3.R",
  "plot_figure_4.R",
  "plot_figure_5.R"
)

run_plot <- function(script) {
  cat("\n=== ", script, " ===\n", sep = "")
  env <- new.env(parent = globalenv())
  sys.source(file.path(R_DIR, script), envir = env)
}

for (s in PLOT_SCRIPTS) run_plot(s)

cat("\nDone.\n")
cat("Panel PNGs:    $CLINICAL_FIGURES_OUT (or /data/.../manuscript_figures/)\n")
cat("Target figures: $CLINICAL_FIGURES_OUT/target_figures/\n")
````
