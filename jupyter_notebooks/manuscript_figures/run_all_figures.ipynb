# Run all manuscript figures

This notebook calls the figure-data prep scripts and plotting scripts with the active kernel Python (`sys.executable`), so it uses the same conda environment as this notebook.

```python
from __future__ import annotations

import subprocess
import sys
from pathlib import Path

def find_repo_root(start: Path | None = None) -> Path:
    start = (start or Path.cwd()).resolve()
    for candidate in (start, *start.parents):
        if (candidate / "python_scripts").is_dir() and (candidate / "jupyter_notebooks").is_dir():
            return candidate
    raise RuntimeError(f"Could not find repo root from {start}")

REPO_ROOT = find_repo_root()
print(f"Repo root: {REPO_ROOT}")
print(f"Python: {sys.executable}")

RUN_PREP = True
RUN_PLOTS = True
RUN_COMPOSE = True

# Optional args for prep_figure_4.py, e.g. ["--decay", "0.1"] or ["--input", "/path/to/file.csv"]
FIGURE4_PREP_ARGS = []
```

```python
PREP_SCRIPTS = [
    ["jupyter_notebooks/manuscript_figures/data_generation/prep_figure_1.py"],
    ["jupyter_notebooks/manuscript_figures/data_generation/prep_figure_2.py"],
    ["jupyter_notebooks/manuscript_figures/data_generation/prep_figure_3.py"],
    ["jupyter_notebooks/manuscript_figures/data_generation/prep_figure_4.py", *FIGURE4_PREP_ARGS],
    ["jupyter_notebooks/manuscript_figures/data_generation/prep_figure_5.py"],
]

PLOT_SCRIPTS = [
    ["jupyter_notebooks/manuscript_figures/plot_figure_1_cohort.py"],
    ["jupyter_notebooks/manuscript_figures/plot_figure_2_text_vs_base.py"],
    ["jupyter_notebooks/manuscript_figures/plot_figure_3_modality_comparison.py"],
    ["jupyter_notebooks/manuscript_figures/plot_figure_4_trajectories.py"],
    ["jupyter_notebooks/manuscript_figures/plot_figure_5_biomarkers.py"],
]

COMPOSE_SCRIPTS = [
    ["jupyter_notebooks/manuscript_figures/compose_target_figures.py"],
]
```

```python
def run_script(args: list[str]) -> None:
    print("\n=== " + " ".join(args) + " ===", flush=True)
    subprocess.run([sys.executable, *args], cwd=REPO_ROOT, check=True)

if RUN_PREP:
    for args in PREP_SCRIPTS:
        run_script(args)

if RUN_PLOTS:
    for args in PLOT_SCRIPTS:
        run_script(args)

if RUN_COMPOSE:
    for args in COMPOSE_SCRIPTS:
        run_script(args)

print("\nDone. Panel PNGs are in jupyter_notebooks/manuscript_figures/figures/", flush=True)
print("Target figures are in jupyter_notebooks/manuscript_figures/figures/target_figures/", flush=True)
```
